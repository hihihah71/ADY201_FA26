```python
pbf_path = Path("../data/raw/osm/vietnam-261014.osm.pbf")
osm = OSM(
    str(pbf_path)
)
```

```python
boundary = osm.get_boundaries(name="Đồng Nai")

print(boundary[["name", "admin_level"]])

boundary.boundary.plot(figsize=(8, 8))
```
